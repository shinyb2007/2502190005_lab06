```python
#JOB-A
raw_lines = [
    "MS-FLG-2026-0041, 24.98",
    " ms-flg-2026-0042, 25.01 ",
    "AL-BRK-2026-0007, 24.95",
    "MS-FLG-2026-0043, 25.04",
    "AL-BRK-2026-0008, 25.12",
    "ms-shf-2026-0101, 24.88",
    "MS-FLG-2026-0044, 24.99",
    "AL-BRK-2026-0009, 25.02",
    "MS-SHF-2026-0102, 25.07",
    "MS-FLG-2026-0045, 24.93"
]
def parse_line(line):
    """Take one raw line and return code, family and diameter."""
    code, dia = line.split(",")
    code = code.strip().upper()
    family = code.split("-")[1]
    return code, family, float(dia)
print(parse_line(raw_lines[0]))
print(parse_line(raw_lines[1]))
print(parse_line(raw_lines[5]))
c, f, d = parse_line(" ms-flg-2026-0099 , 25.0 ")

print(c)
print(f)
print(d)
```

```python
#JOB-B
def build_register(lines):
    """Convert raw lines into a family-wise reading dictionary."""
    reg = {}

    for line in lines:
        code, family, dia = parse_line(line)

        if family not in reg:
            reg[family] = []

        reg[family].append(dia)

    return reg
def family_stats(readings):
    """Return count, mean, minimum and maximum of readings."""
    if len(readings) == 0:
        return (0, 0, 0, 0)

    count = len(readings)
    mean = sum(readings) / count
    minimum = min(readings)
    maximum = max(readings)

    return count, mean, minimum, maximum
regA = build_register(raw_lines)

print(regA)

print("SHIFT A REPORT")

for family, readings in regA.items():
    n, mean, minimum, maximum = family_stats(readings)

    print(
        f"{family} | n={n} | mean={mean:.2f} | "
        f"min={minimum:.2f} | max={maximum:.2f}"
    )
print(family_stats([]))
```

```python
#JOB-C
shift_B = [
    "MS-FLG-2026-0051, 25.03",
    "SS-GER-2026-0021, 24.91",
    "AL-BRK-2026-0012, 25.00",
    "SS-GER-2026-0022, 25.14",
    "MS-FLG-2026-0052, 24.97",
    "AL-BRK-2026-0013, 24.96"
]
regB = build_register(shift_B)

print(regB)

famA = set(regA.keys())
famB = set(regB.keys())

print("Both shifts :", famA & famB)
print("Only A      :", famA - famB)
print("New in B    :", famB - famA)
print("Whole day   :", famA | famB)

mean_A_FLG = family_stats(regA["FLG"])[1]
mean_B_FLG = family_stats(regB["FLG"])[1]

mean_A_BRK = family_stats(regA["BRK"])[1]
mean_B_BRK = family_stats(regB["BRK"])[1]

print(f"FLG drift: {mean_B_FLG - mean_A_FLG:.3f} mm")
print(f"BRK drift: {mean_B_BRK - mean_A_BRK:.3f} mm")
```
